```python
def toHex(num):
    if num == 0:
        return "0"

    num &= 0xFFFFFFFF
    res = ""
    while num > 0:
        num, residual = divmod(num, 16)

        if residual < 10:
            res += str(residual)
        else:
            res += chr(residual + 87)

    return res[::-1]
```

```python
num = 26
toHex(num)
```

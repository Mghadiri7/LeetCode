```python
def shortestBeautifulSubstring(s, k):
    left = 0
    ones = 0
    best_str = ""
    n = len(s)

    for right in range(n):
        if s[right] == '1':
            ones += 1

        while left <= right and (ones > k or s[left] == '0'):
            if s[left] == '1':
                ones -= 1
            left += 1

        if ones == k:
            current_str = s[left:right+1]
            if not best_str:
                best_str = current_str
            elif len(best_str) > len(current_str):
                best_str = current_str
            elif len(current_str) == len(best_str) and current_str < best_str:
                best_str = current_str

    return best_str
```

```python
s = "100011001"
k = 3
shortestBeautifulSubstring(s, k)
```

```python
def maxProfit(prices):
    n = len(prices)
    res = 0
    current = prices[0]

    for i in range(1, n):
        p = prices[i]
        if current > p:
            current = p
        else:
            res += (p - current)
            current = p
            
    return res
```

```python
prices = [7,1,5,3,6,4]
maxProfit(prices)
```

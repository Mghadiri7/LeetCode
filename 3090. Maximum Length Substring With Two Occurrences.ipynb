```python
def maximumLengthSubstring(s):
    n = len(s)
    ans = 0
    left = 0
    freq = {}
    for right in range(n):
        char_right = s[right]
        # freq[char] = freq.get(char, 0) + 1
        if char_right not in freq:
            freq[char_right] = 1
        else:
            freq[char_right] += 1

        while freq[char_right] > 2:
            char_left = s[left]
            freq[char_left] -= 1
            left += 1

        ans = max(ans, right - left + 1)

    return ans
```

```python
s = "bcbbbcba"
maximumLengthSubstring(s)
```

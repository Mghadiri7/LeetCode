```python
def allCellsDistOrder(rows, cols, rCenter, cCenter):
    cells = [[r, c] for r in range(rows) for c in range(cols)]
    cells.sort(key = lambda x: abs(x[0] - rCenter) + abs(x[1] - cCenter))
    return cells
```

```python
rows = 2
cols = 3
rCenter = 1
cCenter = 2
allCellsDistOrder(rows, cols, rCenter, cCenter)
```

```python
def thirdMax(nums):
    max_1 = float('-inf')
    max_2 = float('-inf')
    max_3 = float('-inf')

    for num in nums:
        if num > max_1:
            max_3 = max_2
            max_2 = max_1
            max_1 = num

        elif num > max_2 and num != max_1:
            max_3 = max_2
            max_2 = num

        elif num > max_3 and num != max_1 and num != max_2:
            max_3 = num

    return max_3 if max_3 != float('-inf') else max_1
```

```python
nums = [2,2,3,1]
thirdMax(nums)
```

```python
nums = [1,2,2,5,3,5]
thirdMax(nums)
```

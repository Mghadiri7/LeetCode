```python
from collections import Counter

def minimumPushes(word):
    _dict = Counter(word)
    lst = sorted(_dict.values(), reverse=True)
    n = len(lst)
    res = 0
    for i in range(n):
        if i < 8:
            res += lst[i]
        elif i < 16:
            res += (2 * lst[i])
        elif i < 24:
            res += (3 * lst[i])
        else:
            res += (4 * lst[i])

    return res
```

```python
word = "xyzxyzxyzxyz"
minimumPushes(word)
```

```python
word = "aabbccddeeffgghhiiiiii"
minimumPushes(word)
```

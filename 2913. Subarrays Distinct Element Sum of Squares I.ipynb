```python
def sumCounts(nums):
    n = len(nums)
    res = 0
    
    for i in range(n):
        distinct_numbers = set()
        for j in range(i, n):
            if nums[j] not in distinct_numbers:
                distinct_numbers.add(nums[j])

            count = len(distinct_numbers)
            res += count ** 2

    return res
```

```python
nums = [1,2,1]
sumCounts(nums)
```

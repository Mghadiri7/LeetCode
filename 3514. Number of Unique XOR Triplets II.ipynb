```python
def uniqueXorTriplets(nums):
    MAX_XOR = 2048
    pair = [False] * MAX_XOR
    triple = [False] * MAX_XOR

    n = len(nums)
    for i in range(n):
        for j in range(i, n):
            pair[nums[i] ^ nums[j]] = True

    for x in range(MAX_XOR):
        if not pair[x]:
            continue

        for v in nums:
            triple[x ^ v] = True

    return sum(triple)
```

```python
nums = [6,7,8,9]
uniqueXorTriplets(nums)
```

## First Solution

```python
from collections import Counter

def findTheDifference(s, t):
    freq_s = Counter(s)
    freq_t = Counter(t)
    for char in freq_t:
        if char not in freq_s or freq_t[char] > freq_s[char]:
            return char
```

```python
s = "abcd"
t = "abcde"
findTheDifference(s, t)
```

## Second Solution

```python
def findTheDifference(s, t):
    res = 0
    for c in s + t:
        res ^= ord(c)
    return chr(res)
```

```python
s = "abcd"
t = "abcde"
findTheDifference(s, t)
```

```python
def decodeString(s):
    stack = []
    current_string = ""
    current_num = ""

    for ch in s:
        if '0' <= ch <= '9':
            current_num += ch

        elif ch == "[":
            stack.append((current_string, int(current_num)))
            current_string = ""
            current_num = ""

        elif ch == "]":
            prev_string, num = stack.pop()
            current_string = prev_string + num * current_string

        else:
            current_string += ch

    return current_string
```

```python
s = "3[a]2[bc]"
decodeString(s)
```

```python
s = "3[a2[c]]"
decodeString(s)
```

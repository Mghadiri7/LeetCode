## First Solution

```python
def moveZeroes(nums):
    left, right = 0, 0
    n = len(nums) - 1
    while right < n:
        if nums[left] != 0:
            left += 1
        else:
            right = left + 1
            while nums[right] == 0 and right < n:
                right += 1
            nums[left], nums[right] = nums[right], nums[left]
            left += 1
```

```python
nums = [0,1,0,3,12]
moveZeroes(nums)
print(nums)
```

## Second Solution

```python
def moveZeroes(nums):
    n = len(nums)
    counter = 0
    for i in range(n):
        if nums[i]:
            nums[counter] = nums[i]
            counter += 1
    for i in range(counter, n):
        nums[i] = 0
```

```python
nums = [0,1,0,3,12]
moveZeroes(nums)
print(nums)
```

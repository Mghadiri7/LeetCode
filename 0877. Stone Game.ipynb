```python
from functools import cache

def stoneGame(piles):
    if len(piles) % 2 == 0:
        return True

    @cache
    def score(left, right):
        if left == right:
            return piles[left]

        pick_left = piles[left] - score(left+1, right)
        pick_right = piles[right] - score(left, right-1)
        return max(pick_left, pick_right)

    return score(0, len(piles)-1) >= 0
```

```python
piles = [5,3,4,5]
stoneGame(piles)
```

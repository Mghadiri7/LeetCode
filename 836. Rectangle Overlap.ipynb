```python
def isRectangleOverlap(rec1, rec2):
    if rec1[2] <= rec2[0]:
        return False
    if rec1[0] >= rec2[2]:
        return False
    if rec1[3] <= rec2[1]:
        return False
    if rec1[1] >= rec2[3]:
        return False

    return True
```

```python
rec1 = [0,0,2,2]
rec2 = [1,1,3,3]
isRectangleOverlap(rec1, rec2)
```

```python
rec1 = [0,0,1,1]
rec2 = [1,0,2,1]
isRectangleOverlap(rec1, rec2)
```

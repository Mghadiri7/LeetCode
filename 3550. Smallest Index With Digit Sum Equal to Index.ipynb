```python
def smallestIndex(nums):
    n = len(nums)
    for i in range(n):
        total = 0
        num = nums[i]
        while num > 0:
            num, digit = divmod(num, 10)
            total += digit

        if i == total:
            return i

    return -1
```

```python
nums = [1,3,2]
smallestIndex(nums)
```

```python
def largestSumAfterKNegations(nums, k):
    n = len(nums)
    nums.sort()
    for i in range(n):
        if k > 0 and nums[i] < 0:
            nums[i] = -nums[i]
            k -= 1

    if k % 2 == 1:
        min_val = min(nums)
        return sum(nums) - 2 * min_val

    return sum(nums)
```

```python
nums = [2,-3,-1,5,-4]
k = 2
largestSumAfterKNegations(nums, k)
```

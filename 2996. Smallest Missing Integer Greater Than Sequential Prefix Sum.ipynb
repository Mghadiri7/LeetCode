```python
def missingInteger(nums):
    n = len(nums)
    longest = 0
    flag = True
    for i in range(1, n):
        if nums[i] != nums[i-1] + 1:
            longest = i
            flag = False
            break
        longest = i
        
    if flag:
        _sum = sum(nums)
    else:
        _sum = sum(nums[:longest])
        
    while True:
        if _sum not in nums:
            return _sum
        _sum += 1
```

```python
nums = [3,4,5,1,12,14,13]
missingInteger(nums)
```

```python
nums = [29,30,31,32,33,34,35,36,37]
missingInteger(nums)
```

## First Solution

```python
def minMoves(nums):
    _min = min(nums)
    res = 0
    for num in nums:
        res += num - _min

    return res
```

```python
nums = [1,2,3]
minMoves(nums)
```

## Second Solution

```python
def minMoves(nums):
    return sum(nums) - min(nums) * len(nums)
```

```python
nums = [1,2,3]
minMoves(nums)
```

```python
def maxValue(nums):
    n = len(nums)
    if not n:
        return []

    suffix = [0] * n
    suffix[-1] = nums[-1]
    for i in range(n-2, -1, -1):
        suffix[i] = min(suffix[i+1], nums[i])

    ans = [0] * n
    current_max = nums[0]
    start = 0

    for i in range(n):
        current_max = max(current_max, nums[i])

        if i == n-1 or current_max <= suffix[i + 1]:
            for j in range(start, i + 1):
                ans[j] = current_max

            start = i + 1
            if i + 1 < n:
                current_max = nums[i + 1]

    return ans
```

```python
nums = [2,3,1]
maxValue(nums)
```

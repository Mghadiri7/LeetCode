```python
def intersection(nums1, nums2):
    ans = []
    nums1_set = set(nums1)
    nums2_set = set(nums2)
    if len(nums1_set) < len(nums2_set):
        for num in nums1_set:
            if num in nums2_set:
                ans.append(num)
    else:
        for num in nums2_set:
            if num in nums1_set:
                ans.append(num)

    return ans
```

```python
nums1 = [4,9,5]
nums2 = [9,4,9,8,4]
intersection(nums1, nums2)
```

```python
def reverseVowels(s):
    s_list = list(s)
    _set = {"a", "e", "i", "o", "u", "A", "E", "I", "O", "U"}
    left = 0
    right = len(s) - 1
    while left < right:
        if s_list[left] in _set and s_list[right] in _set:
            s_list[left], s_list[right] = s_list[right], s_list[left]
            left += 1
            right -= 1
        elif s_list[left] in _set:
            right -= 1
        elif s_list[right] in _set:
            left += 1
        else:
            left += 1
            right -= 1

    return "".join(s_list)
```

```python
s = "IceCreAm"
reverseVowels(s)
```

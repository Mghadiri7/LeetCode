```python
def containsNearbyDuplicate(nums, k):
    seen = {}

    for index, value in enumerate(nums):
        if value in seen and index - seen[value] <= k:
            return True
        seen[value] = index

    return False
```

```python
nums = [1,2,3,1]
k = 3
containsNearbyDuplicate(nums, k)
```

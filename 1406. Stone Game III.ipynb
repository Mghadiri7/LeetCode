```python
def stoneGameIII(stoneValue):
    n = len(stoneValue)
    dp = [float('-inf')] * n + [0]
    
    for i in range(n-1, -1, -1):
        take = 0
        for j in range(1, 4):
            if i + j <= n:
                take += stoneValue[i+j-1]
                dp[i] = max(dp[i], take - dp[i + j])
        
    if dp[0] > 0:
        return "Alice"
    elif dp[0] < 0:
        return "Bob"
    else:
        return "Tie"
```

```python
stoneValue = [1,2,3,7]
stoneGameIII(stoneValue)
```

```python
stoneValue = [1,2,3,-9]
stoneGameIII(stoneValue)
```

```python
def checkOverlap(radius, xCenter, yCenter, x1, y1, x2, y2):
    x = max(x1, min(xCenter, x2)) - xCenter
    y = max(y1, min(yCenter, y2)) - yCenter

    return x * x + y * y <= radius * radius
```

```python
radius = 1
xCenter = 0
yCenter = 0
x1 = 1
y1 = -1
x2 = 3
y2 = 1
checkOverlap(radius, xCenter, yCenter, x1, y1, x2, y2)
```

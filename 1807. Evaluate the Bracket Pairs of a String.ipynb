```python
def evaluate(s, knowledge):
    knowledge_dict = {key: val for key, val in knowledge}

    in_bracket = False
    current_key = []
    res = []
    
    for ch in s:
        if ch == "(":
            in_bracket = True
            
        elif ch == ")":
            in_bracket = False
            key_str = "".join(current_key)
            res.append(knowledge_dict.get(key_str, "?"))
            current_key = []

        else:
            if in_bracket:
                current_key.append(ch)
            else:
                res.append(ch)

    return "".join(res)
```

```python
s = "(name)is(age)yearsold"
knowledge = [["name","bob"],["age","two"]]
evaluate(s, knowledge)
```

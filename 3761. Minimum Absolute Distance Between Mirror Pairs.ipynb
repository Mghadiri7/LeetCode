```python
def minMirrorPairDistance(nums):
    ans = float('inf')
    seen = {}
    for index, value in enumerate(nums):
        if value in seen:
            ans = min(ans, index - seen[value])

        reverse = int(str(value)[::-1])
        seen[reverse] = index

    return ans if ans != float('inf') else -1
```

```python
nums = [12,21,45,33,54]
minMirrorPairDistance(nums)
```

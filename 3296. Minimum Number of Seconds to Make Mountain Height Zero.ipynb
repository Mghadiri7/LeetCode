```python
import math

def minNumberOfSeconds(mountainHeight, workerTimes):
    def finish(time_limit):
        total_reduced = 0
        for t in workerTimes:
            max_x = int((-1 + math.sqrt(1 + 8 * time_limit / t)) // 2)
            total_reduced += max_x
            
            if total_reduced >= mountainHeight:
                return True
                
        
    left = 1
    max_worker_time = max(workerTimes)
    right = max_worker_time * (mountainHeight * (mountainHeight + 1)) // 2
    ans = right
    
    while left <= right:
        mid = (left + right) // 2
        if finish(mid):
            ans = mid
            right = mid - 1
        else:
            left = mid + 1
            
    return ans
```

```python
mountainHeight = 10
workerTimes = [3,2,2,4]
minNumberOfSeconds(mountainHeight, workerTimes)
```

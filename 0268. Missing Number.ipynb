## First solution

```python
def missingNumber(nums):
    n = len(nums)
    if n in nums:
        nums.sort()
        for index, value in enumerate(nums):
            if index != value:
                return index
    return n
```

```python
nums = [9,6,4,2,3,5,7,0,1]
missingNumber(nums)
```

## Second solution

```python
def missingNumber(nums):
    _sum = sum(nums)
    n = len(nums)
    return (n * (n + 1)) // 2 - _sum
```

```python
nums = [9,6,4,2,3,5,7,0,1]
missingNumber(nums)
```

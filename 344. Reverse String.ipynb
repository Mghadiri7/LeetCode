## First Solution

```python
def reverseString(s):
    n = len(s) // 2
    for i in range(n):
        s[i], s[-1 - i] = s[-1 - i], s[i]
    return s
```

```python
s = ["H","a","n","n","a","h","p"]
reverseString(s)
```

## Second Solution

```python
def reverseString(s):
    left = 0
    right = len(s) - 1
    while left < right:
        s[left], s[right] = s[right], s[left]
        left += 1
        right -= 1
    return s
```

```python
s = ["H","a","n","n","a","h","p"]
reverseString(s)
```

## First Solution

```python
def minPartitions(n):
    res = 0
    for s in n:
        if int(s) > res:
            res = int(s)
    return res
```

```python
n = "27346209830709182346"
minPartitions(n)
```

## Second Solution

```python
def minPartitions(n):
    for i in range(9, 0, -1):
        if str(i) in n:
            return i
    return 0
```

```python
n = "27346209830709182346"
minPartitions(n)
```

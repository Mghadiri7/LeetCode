## First Solution

```python
from collections import Counter

def totalNumbers(digits):
    _dict = Counter(digits)
    ans = 0

    for num in range(100, 1000, 2):
        d1 = num // 100
        d2 = (num // 10) % 10
        d3 = num % 10

        _dict[d1] -= 1
        _dict[d2] -= 1
        _dict[d3] -= 1

        if _dict[d1] >= 0 and _dict[d2] >= 0 and _dict[d3] >= 0:
            ans += 1

        _dict[d1] += 1
        _dict[d2] += 1
        _dict[d3] += 1

    return ans
```

```python
digits = [1,2,3,4]
totalNumbers(digits)
```

## Second Solution

```python
from collections import Counter

def totalNumbers(digits):
    freq = Counter(digits)
    ans = 0

    for d1 in range(1, 10):
        if freq[d1] > 0:
            freq[d1] -= 1
            for d2 in range(10):
                if freq[d2] > 0:
                    freq[d2] -= 1
                    for d3 in [0,2,4,6,8]:
                        if freq[d3] > 0:
                            ans += 1
                        
                    freq[d2] += 1
            freq[d1] += 1

    return ans
```

```python
digits = [1,2,3,4]
totalNumbers(digits)
```

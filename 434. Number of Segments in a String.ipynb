## Second Solution

```python
def countSegments(s):
    if not s:
        return 0
        
    res = 0
    n = len(s)
    if s[0] != ' ':
        res += 1

    for i in range(1, n):
        if s[i] != ' ' and s[i-1] == ' ':
            res += 1

    return res
```

```python
s = "Hello, my name is John"
countSegments(s)
```

## First Solution

```python
def minAddToMakeValid(s):
    stack = []
    for ch in s:
        if ch == "(":
            stack.append(ch)
        else:
            if stack and stack[-1] == "(":
                stack.pop()
            else:
                stack.append(ch)

    return len(stack)
```

```python
s = "())"
minAddToMakeValid(s)
```

## Second Solution

```python
def minAddToMakeValid(s):
    open_p = 0
    close_p = 0
    for ch in s:
        if ch == "(":
            open_p += 1
        else:
            if open_p > 0:
                open_p -= 1
            else:
                close_p += 1

    return open_p + close_p
```

```python
s = "())"
minAddToMakeValid(s)
```

```python
s = "()))(("
minAddToMakeValid(s)
```

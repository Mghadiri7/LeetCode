```python
def sumGame(num):
    n = len(num)
    half = n // 2
    sum_left, sum_right = 0, 0
    q_left, q_right = 0, 0

    for i in range(half):
        if num[i] == '?':
            q_left += 1
        else:
            sum_left += int(num[i])

    for i in range(half, n):
        if num[i] == '?':
            q_right += 1
        else:
            sum_right += int(num[i])

    if (q_left + q_right) % 2 == 1:
        return True

    if 2 * (sum_left - sum_right) == 9 * (q_right - q_left):
        return False

    return True
```

```python
num = "?3295???"
sumGame(num)
```

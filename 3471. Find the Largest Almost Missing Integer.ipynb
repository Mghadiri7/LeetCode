```python
from collections import Counter

def largestInteger(nums, k):
    n = len(nums)
    if k == 1:
        _dict = Counter(nums)
        ans = -1
        for key, value in _dict.items():
            if value == 1:
                ans = max(ans, key)

        return ans

    elif k == n:
        return max(nums)

    ans = -1
    if nums[0] not in nums[1:]:
        ans = nums[0]
    if nums[-1] not in nums[:n-1]:
        ans = max(ans, nums[-1])
    return ans
```

```python
nums = [3,9,2,1,7]
k = 3
largestInteger(nums, k)
```

```python
nums = [0,0]
k = 1
largestInteger(nums, k)
```

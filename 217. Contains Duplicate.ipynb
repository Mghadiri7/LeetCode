## First solution

```python
def containsDuplicate(nums):
    nums.sort()
    n = len(nums) - 1
    for i in range(n):
        if nums[i] == nums[i+1]:
            return True
    return False
```

```python
nums = [1,2,3,1]
containsDuplicate(nums)
```

## Second Solution

```python
def containsDuplicate(nums):
    return len(set(nums)) != len(nums)
```

```python
nums = [1,2,3,4]
containsDuplicate(nums)
```

```python
def resultArray(nums):
    n = len(nums)
    arr1, arr2 = [], []
    arr1.append(nums[0])
    arr2.append(nums[1])
    for i in range(2, n):
        if arr1[-1] > arr2[-1]:
            arr1.append(nums[i])
        else:
            arr2.append(nums[i])
            
    return arr1 + arr2
```

```python
nums = [5,4,3,8]
resultArray(nums)
```

```python
def maxDepthAfterSplit(seq):
    depth = 0
    res = []

    for ch in seq:
        if ch == "(":
            depth += 1
            res.append(depth % 2)

        else:
            res.append(depth % 2)
            depth -= 1

    return res
```

```python
seq = "()(())()"
maxDepthAfterSplit(seq)
```

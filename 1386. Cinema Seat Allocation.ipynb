```python
import collections

def maxNumberOfFamilies(n, reservedSeats):
    seats = collections.defaultdict(set)

    for row, seat in reservedSeats:
        if 2 <= seat <= 9:
            seats[row].add(seat)
    
    ans = 2 * n
    
    for row, reserved in seats.items():
        ans -= 2
        left_free = not reserved.intersection({2, 3, 4, 5})
        right_free = not reserved.intersection({6, 7, 8, 9})
        middle_free = not reserved.intersection({4, 5, 6, 7})

        if left_free and right_free:
            ans += 2
        elif left_free or right_free or middle_free:
            ans += 1

    return ans
```

```python
n = 3
reservedSeats = [[1,2],[1,3],[1,8],[2,6],[3,1],[3,10]]
maxNumberOfFamilies(n, reservedSeats)
```

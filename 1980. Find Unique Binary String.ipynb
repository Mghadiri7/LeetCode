```python
def findDifferentBinaryString(nums):
    n = len(nums)
    ans = "0" * n
    for i in range(n + 1):
        if ans not in nums:
            return ans
        ans = ans[:i] + "1" + ans[i+1:]
```

```python
nums = ["111","011","001"]
findDifferentBinaryString(nums)
```

```python
nums = ["00","01"]
findDifferentBinaryString(nums)
```

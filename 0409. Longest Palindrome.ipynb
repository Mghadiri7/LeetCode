```python
from collections import Counter

def longestPalindrome(s):
    freq = Counter(s)
    max_odd = 0
    ans = 0
    flag = False
    for value in freq.values():
        if value % 2 == 0:
            ans += value
        else:
            ans += value - 1
            flag = True

    return ans + 1 if flag == True else ans
```

```python
s = "abccccdd"
longestPalindrome(s)
```

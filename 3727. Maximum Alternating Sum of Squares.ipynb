## First Solution

```python
def maxAlternatingSum(nums):
    n = len(nums)
    res = 0
    for i in range(n):
        temp = nums[i]
        if temp < 0:
            nums[i] = abs(temp)

    nums.sort(reverse=True)
    
    if n % 2 != 0:
        res += nums[n // 2] ** 2
        half = n // 2
        for i in range(half):
            res += nums[i] ** 2
            res -= nums[i + half + 1] ** 2

    else:
        half = n // 2
        for i in range(half):
            res += nums[i] ** 2
            res -= nums[i + half] ** 2

    return res
```

```python
nums = [1,-1,2,-2,3,-3]
maxAlternatingSum(nums)
```

## Second Solution

```python
def maxAlternatingSum(nums):
    lst = sorted((x * x for x in nums), reverse=True)
    half = (len(lst) + 1) // 2
    return sum(lst[:half]) - sum(lst[half:])
```

```python
nums = [1,-1,2,-2,3,-3]
maxAlternatingSum(nums)
```

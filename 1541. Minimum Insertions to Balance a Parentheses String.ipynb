```python
def minInsertions(s):
    insert = 0
    need_right = 0
    for ch in s:
        if ch == "(":
            if need_right % 2 != 0:
                insert += 1
                need_right -= 1
            need_right += 2
            
        else:
            need_right -= 1
            if need_right < 0:
                insert += 1
                need_right = 1

    return insert + need_right
```

```python
s = "))())("
minInsertions(s)
```

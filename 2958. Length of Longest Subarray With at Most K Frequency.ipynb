```python
def maxSubarrayLength(nums, k):
    n = len(nums)
    ans = 0
    freq = {}
    left = 0
    for right in range(n):
        current = nums[right]
        freq[current] = freq.get(current, 0) + 1

        while freq[current] > k:
            left_num = nums[left]
            freq[left_num] -= 1
            left += 1

        ans = max(ans, right - left + 1)

    return ans
```

```python
nums = [1,2,3,1,2,3,1,2]
k = 2
maxSubarrayLength(nums, k)
```

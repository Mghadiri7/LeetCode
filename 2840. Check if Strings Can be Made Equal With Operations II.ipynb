```python
from collections import Counter

def checkStrings(s1, s2):
    even_s1 = Counter(s1[::2])
    odd_s1 = Counter(s1[1::2])
    even_s2 = Counter(s2[::2])
    odd_s2 = Counter(s2[1::2])
    if even_s1 != even_s2 or odd_s1 != odd_s2:
        return False
    return True
```

```python
s1 = "abcdba"
s2 = "cabdab"
checkStrings(s1, s2)
```

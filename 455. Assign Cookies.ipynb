```python
def findContentChildren(g, s):
    g.sort()
    s.sort()
    n = len(g)
    m = len(s)
    res = 0
    i = 0
    j = 0

    while i < n and j < m:
        if g[i] <= s[j]:
            res += 1
            i += 1
            j += 1
        else:
            j += 1

    return res
```

```python
g = [1,2]
s = [1,2,3]
findContentChildren(g, s)
```

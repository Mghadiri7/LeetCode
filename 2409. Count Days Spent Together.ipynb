```python
def countDaysTogether(arriveAlice, leaveAlice, arriveBob, leaveBob):
    days_in_months = [31, 28, 31, 30, 31, 30, 31, 31, 30, 31, 30, 31]

    def date_to_days(date_str):
        month = int(date_str[:2])
        day = int(date_str[3:])

        return sum(days_in_months[:month - 1]) + day
    
    arrive_A = date_to_days(arriveAlice)
    leave_A = date_to_days(leaveAlice)
    arrive_B = date_to_days(arriveBob)
    leave_B = date_to_days(leaveBob)
    
    start = max(arrive_A, arrive_B)
    end = min(leave_A, leave_B)

    return max(0, end - start + 1)
```

```python
arriveAlice = "08-15"
leaveAlice = "08-18"
arriveBob = "08-16"
leaveBob = "08-19"
countDaysTogether(arriveAlice, leaveAlice, arriveBob, leaveBob)
```

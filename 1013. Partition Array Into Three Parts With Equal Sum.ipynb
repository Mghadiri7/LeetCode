```python
def canThreePartsEqualSum(arr):
    _sum = sum(arr)
    if _sum % 3 != 0:
        return False
        
    one_third = _sum // 3
    n = len(arr)
    add = 0
    check = 0
    for i in range(n):
        add += arr[i]
        if add == one_third:
            add = 0
            check += 1
            if check == 3:
                return True
            
    return False
```

```python
arr = [0,2,1,-6,6,-7,9,1,2,0,1]
canThreePartsEqualSum(arr)
```

```python
arr = [3,3,6,5,-2,2,5,1,-9,4]
canThreePartsEqualSum(arr)
```

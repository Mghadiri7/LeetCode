## First Solution

```python
def arrangeCoins(n):
    res = 0
    while res * (res + 1) / 2 < n:
        res += 1

    return res if res * (res + 1) / 2 == n else res - 1
```

```python
n = 8
arrangeCoins(n)
```

## Third Solution

```python
def arrangeCoins(n):
    left = 1
    right = n

    while left <= right:
        mid = (left + right) // 2
        needed = mid * (mid + 1) // 2

        if needed == n:
            return mid

        elif needed > n:
            right = mid - 1

        else:
            left = mid + 1

    return right
```

```python
n = 8
arrangeCoins(n)
```

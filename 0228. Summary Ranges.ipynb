```python
def summaryRanges(nums):
    if nums == []:
        return []
        
    ans = []
    n = len(nums) - 1
    flag = False
    _str = str(nums[0])
    for i in range(n):
        if nums[i] + 1 == nums[i+1]:
            flag = True
            continue

        if flag:
            _str = _str + '->' + str(nums[i])
            
        ans.append(_str)
        _str = str(nums[i+1])
        flag = False

    if flag:
        _str = _str + '->' + str(nums[-1])     
    ans.append(_str)
        
    return ans
```

```python
nums = [0,1,2,4,5,7]
summaryRanges(nums)
```

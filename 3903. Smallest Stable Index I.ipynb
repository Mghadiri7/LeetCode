```python
def firstStableIndex(nums, k):
    _max = nums[0]
    _min = min(nums)
    n = len(nums) - 1
    for index, value in enumerate(nums):
        if value > _max:
            _max = value
        if _max - _min <= k:
            return index
        if value == _min and index < n:
            _min = min(nums[index+1:])
            
    return -1
```

```python
nums = [5,0,1,4]
k = 3
firstStableIndex(nums, k)
```

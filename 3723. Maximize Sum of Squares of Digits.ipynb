```python
def maxSumOfSquares(num, _sum):
    res = ""
    if 9 * num < _sum:
        return res

    for i in range(num):
        if _sum > 8:
            _sum -= 9
            res += "9"
        elif _sum > 0:
            res += str(_sum)
            _sum = 0
        else:
            res += "0"
    
    return res
```

```python
num = 2
_sum = 17
maxSumOfSquares(num, _sum)
```

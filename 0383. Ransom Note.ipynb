```python
from collections import Counter

def canConstruct(ransomNote, magazine):
    freq = Counter(magazine)
    for char in ransomNote:
        if char not in freq or freq[char] < 1:
            return False
        freq[char] -= 1
        
    return True
```

```python
ransomNote = "aa"
magazine = "aab"
canConstruct(ransomNote, magazine)
```

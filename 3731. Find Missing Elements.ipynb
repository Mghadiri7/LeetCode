```python
def findMissingElements(nums):
    res = []
    nums.sort()
    _min = nums[0] + 1
    _max = nums[-1]
    i = 1
    for x in range(_min, _max):
        if x == nums[i]:
            i += 1
        else:
            res.append(x)

    return res
```

```python
nums = [1,4,2,5]
findMissingElements(nums)
```

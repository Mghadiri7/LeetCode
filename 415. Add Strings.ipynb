```python
def addStrings(num1, num2):
    n = len(num1) - 1
    m = len(num2) -1 
    carry = 0
    res = []
    
    while n >= 0 or m >= 0 or carry > 0:
        d1 = ord(num1[n]) - ord("0") if n >= 0 else 0
        d2 = ord(num2[m]) - ord("0") if m >= 0 else 0

        current = d1 + d2 + carry
        carry, digit = divmod(current, 10)

        res.append(str(digit))
        n -= 1
        m -= 1

    return "".join(res[::-1])
```

```python
num1 = "456"
num2 = "77"
addStrings(num1, num2)
```

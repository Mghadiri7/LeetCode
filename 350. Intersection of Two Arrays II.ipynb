```python
from collections import Counter

def intersect(nums1, nums2):
    ans = []
    dict_nums1 = Counter(nums1)
    dict_nums2 = Counter(nums2)
    if len(dict_nums1) < len(dict_nums2):
        for key, value in dict_nums1.items():
            if key in dict_nums2:
                for i in range(min(value, dict_nums2[key])):
                    ans.append(key)
    else:
        for key, value in dict_nums2.items():
            if key in dict_nums1:
                for i in range(min(value, dict_nums1[key])):
                    ans.append(key)

    return ans
```

```python
nums1 = [1,2,2,1]
nums2 = [2,2]
intersect(nums1, nums2)
```

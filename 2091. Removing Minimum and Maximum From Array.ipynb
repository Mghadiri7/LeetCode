```python
def minimumDeletions(nums):
    min_index = nums.index(min(nums))
    max_index = nums.index(max(nums))
    n = len(nums)
    front = min(min_index, max_index)
    back = max(min_index, max_index)
    return min(back + 1, n - front, front + n - back + 1)
```

```python
nums = [2,10,7,5,4,1,8,6]
minimumDeletions(nums)
```

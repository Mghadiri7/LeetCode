```python
def scoreOfParentheses(s):
    n = len(s)
    res = 0
    depth = 0
    for i in range(n):
        if s[i] == "(":
            depth += 1

        else:
            depth -= 1
            if s[i-1] == "(":
                res += 1 << depth
                
    return res      
```

```python
s = "(())"
scoreOfParentheses(s)
```

```python
s = "(()(()))"
scoreOfParentheses(s)
```

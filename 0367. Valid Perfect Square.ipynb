## First Solution

```python
def isPerfectSquare(num):
    s = 1
    squared = 1
    while squared <= num:
        if num == squared:
            return True
        s += 1
        squared = s * s
    return False
```

```python
num = 16
isPerfectSquare(num)
```

## Second Solution

```python
def isPerfectSquare(num):
    L = 0
    R = num
    while L <= R:
        middle = (R + L) // 2
        if middle * middle == num:
            return True
        elif middle * middle > num:
            R = middle - 1
        else:
            L = middle + 1
    return False
```

```python
num = 16
isPerfectSquare(num)
```

```python
def numSteps(s):
    n = len(s) - 1
    step = 0
    carry = 0
    
    for i in range(n, 0, -1):
        value = int(s[i]) + carry
        
        if value == 1:
            step += 2
            carry = 1

        else:
            step += 1
        
    return step + carry
```

```python
s = "1101"
numSteps(s)
```

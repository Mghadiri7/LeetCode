## First Solution

```python
def addDigits(num):
    while num > 9:
        temp = num % 10
        num //= 10
        num += temp
    return num
```

```python
num = 38
addDigits(num)
```

## Second solution

```python
def addDigits(num):
    if num == 0:
        return 0
    return 1 + (num - 1) % 9
```

```python
num = 38
addDigits(num)
```

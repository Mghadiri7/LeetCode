```python
def removeOuterParentheses(s):
    ans = ""
    counter = 0
    for _ in s:
        if _ == '(':
            if counter > 0:
                ans += '('
            counter += 1
        else:
            counter -= 1
            if counter > 0:
                ans += ')'

    return ans
```

```python
s = "(()())(())(()(()))"
removeOuterParentheses(s)
```

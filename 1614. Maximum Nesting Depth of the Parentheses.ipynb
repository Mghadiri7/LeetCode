```python
def maxDepth(s):
    ans = 0
    current = 0
    n = len(s)
    for i in range(n):
        if s[i] == "(":
            current += 1
        elif s[i] == ")":
            ans = max(ans, current)
            current -= 1
    
    return ans
```

```python
s = "(1+(2*3)+((8)/4))+1"
maxDepth(s)
```

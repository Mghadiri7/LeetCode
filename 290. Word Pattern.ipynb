```python
def wordPattern(pattern, s):
    words = s.split()
    if len(pattern) != len(words):
        return False

    char_to_word = {}
    mapped_word = set()

    for char, word in zip(pattern, words):
        if char in char_to_word:
            if char_to_word[char] != word:
                return False

        elif word in mapped_word:
            return False

        else:
            char_to_word[char] = word
            mapped_word.add(word)

    return True
```

```python
pattern = "abba"
s = "dog cat cat dog"
wordPattern(pattern, s)
```

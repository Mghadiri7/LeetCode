```python
def minFlips(s):
    n = len(s)
    s += s
    diff_0 = 0
    diff_1 = 0
    ans = float('inf')
    for i in range(len(s)):
        char1 = '0' if i % 2 == 0 else '1'
        char2 = '1' if i % 2 == 0 else '0'

        if s[i] != char1: diff_0 += 1
        if s[i] != char2: diff_1 += 1

        if i >= n:
            left_index = i - n
            left_char = s[left_index]

            left_char1 = '0' if left_index % 2 == 0 else '1'
            left_char2 = '1' if left_index % 2 == 0 else '0'

            if left_char != left_char1: diff_0 -= 1
            if left_char != left_char2: diff_1 -= 1

        if i >= n - 1:
            ans = min(ans, diff_0, diff_1)

    return ans
```

```python
s = "111000"
minFlips(s)
```
